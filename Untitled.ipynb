```python
def start_end_decorator(func):
    def wrapper(*args, **kwargs):
        print("start")
        result = func(*args,**kwargs)
        print("end")
        return result
    return wrapper

@start_end_decorator
def name():
    print("Alex")

@start_end_decorator
def l(x):
    return x+5

name()
l(4)
```
